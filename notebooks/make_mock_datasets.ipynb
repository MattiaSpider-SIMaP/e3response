```python
import zipfile
import os
from pathlib import Path
```

```python
SOURCE_DIR = Path("/home/mattia/Desktop/MLcodes/e3response/data/qm9_nmr") 
DEST_DIR = Path("/home/mattia/Desktop/MLcodes/e3response/test/mock_datasets/qm9_nmr")
DEST_DIR.mkdir(parents=True, exist_ok=True)

N_LOGS = 5

DATASETS = [
    "gasphase",
    "DMSO",
    "methanol",
    "THF",
    "CCl4",
    "acetone",
]

for name in DATASETS:
    src_zip = SOURCE_DIR / f"QM9nmr_{name}_logs.zip"
    dest_zip = DEST_DIR / f"QM9nmr_{name}_logs.zip"

    if not src_zip.exists():
        print(f"⚠️  Missing source file: {src_zip}")
        continue

    print(f"Processing {src_zip.name} → extracting {N_LOGS} logs...")

    with zipfile.ZipFile(src_zip, "r") as zsrc:
        log_files = [f for f in zsrc.namelist() if f.endswith(".log")][:N_LOGS]

        if not log_files:
            print(f"❌ No log files found in {src_zip}")
            continue

        with zipfile.ZipFile(dest_zip, "w", compression=zipfile.ZIP_DEFLATED) as zdst:
            for log_file in log_files:
                with zsrc.open(log_file) as src:
                    content = src.read()
                    zdst.writestr(log_file, content)

    print(f"✅ Created {dest_zip} ({os.path.getsize(dest_zip)/1e6:.2f} MB)")
```
